```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pickle
```

```python
df['combined']=(
    df['song_name'].astype(str)+' '+
    df['artist'].astype(str)+' '+
    df['album'].astype(str)+' '+
    df['genre'].astype(str)
)
```

```python
df['combined']
```

```python
tfidf = TfidfVectorizer(stop_words='english')
vectors = tfidf.fit_transform(df['combined'])
```

```python
similarity = cosine_similarity(vectors)
```

```python
def recommend(song):
    song = song.lower()

    try:
        index =  df[df['song'].str.lower()==song].index[0]
    except:
        return []
    distances = similarity[index]

    song_list = sorted(list(enumerate(distances)),
                       reverse=True,key=lambda x: x[1])[1:6]
    recommend_song=[]

    for i in song_list:
        recommend_song.append(df.iloc[i[0]]['song_name'])
    return recommend_song
```

```python
pickle.dump(df,open("song_df.pkl","wb"))
pickle.dump(similarity,open("similarity.pkl","wb"))
pickle.dump(tfidf,open("tfidf.pkl","wb"))
```
